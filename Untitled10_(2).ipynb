## 🛠️ Step 4: Data Preprocessing

> **Why preprocessing?**  
> Naïve Bayes (GaussianNB) works with **numeric values only**.  
> We need to convert categorical columns (text) into numbers using **Label Encoding/ oneHot Encoding**.

```python
df
df_copy = df.copy()
# df_copy = df_copy.drop('applicant_id', axis = 1)
```

```python
# Identify categorical columns
cat_col = df.select_dtypes(include=['object', 'string']).columns
df_copy = pd.get_dummies(df, columns=cat_col, dtype=int, drop_first=True)
```

```python
# one hot encoding
df_copy = df.copy()
df_copy = pd.get_dummies(df_copy,
                         columns = cat_col,
                         dtype = int,
                         drop_first= True)
df_copy.head(2)
df_copy.shape
```

```python
from sklearn.preprocessing import LabelEncoder
Le = LabelEncoder()
df_copy['loan_status'] = Le.fit_transform(df_copy['loan_status'])
df_copy['loan_status'].head()
```

```python
# Define Features (X) and Target (y)
X = df_copy.drop('loan_status', axis=1)
y = df_copy['loan_status']
```

## ✂️ Step 5: Train-Test Split

> We split the data into **80% training** and **20% testing**.  
> The model **learns from training data** and is **evaluated on test data** it has never seen.

```python
# Train-Test Split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=.2, random_state=42)
```

```python
# Perform class imbalance
from imblearn.over_sampling import SMOTE
sm = SMOTE(random_state=42)
X_train, y_train = sm.fit_resample(X_train, y_train)
X_train.shape, y_train.shape
```

## 🤖 Step 6: Train the Naïve Bayes Model

> **GaussianNB** assumes features follow a **Normal (Gaussian) distribution**.  
> It calculates the **mean** and **variance** of each feature for each class.

```python
# Initialize and Train the model
from sklearn.naive_bayes import GaussianNB
gnb = GaussianNB()
gnb.fit(X_train, y_train)
```

```python
y_pred = gnb.predict(X_test)
y_pred[:10]
```

```python
compare = pd.DataFrame({
    'Actual' : y_test,
    'Predicted' : y_pred
})
compare.head(10)
```

## 📊 Step 7: Model Evaluation

```python
from sklearn.metrics import( precision_score, recall_score, f1_score, accuracy_score)
```

```python
print("Accuracy :", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall   :", recall_score(y_test, y_pred))
print("F1 Score :", f1_score(y_test, y_pred))
```

## 🔮 Step 8: Predict for a New Loan Applicant

> Let's simulate a **real-world prediction** for a new applicant who has **not been seen by the model** before.

```python
for c in cat_col:
    print(c, df[c].unique())
```

```python
new_applicant = pd.DataFrame([{
    'person_age': 22,
    'person_gender': 'female',
    'person_education': 'High School',
    'person_income': 20000,
    'person_emp_exp': 0,
    'person_home_ownership': 'RENT',
    'loan_amnt': 15000,
    'loan_intent': 'MEDICAL',
    'loan_int_rate': 16.5,
    'loan_percent_income': 0.75,
    'cb_person_cred_hist_length': 2,
    'credit_score': 520,
    'previous_loan_defaults_on_file': 'Yes'
}])
```

```python
new_encoded = pd.get_dummies(new_applicant, columns=cat_col, dtype=int)
new_encoded = new_encoded.reindex(columns=X.columns, fill_value=0)

pred = gnb.predict(new_encoded)[0]
proba = gnb.predict_proba(new_encoded)[0]

print("Prediction:", pred)
print(f"Probability of class 0: {proba[0]:.2%}")
print(f"Probability of class 1: {proba[1]:.2%}")
```
